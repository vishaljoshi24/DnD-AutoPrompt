```python
from datasets import load_dataset
import pandas as pd
```

```python
df = pd.read_json("hf://datasets/garrykuwanto/crd3_training_pairs/crd3_training_pairs_no_llm.json")
```

```python
df
```

```python
query = []
passage = []
for i in range(len(df[0:])):
  query.append(df['query'][i])
  passage.append(df['passage'][i])

df1 = pd.DataFrame({'context': query, 'current turn': passage})
```

```python
df1
```

```python
df2 = df1[(~df1['context'].str.startswith('Scene context:')) & (~df1['context'].str.startswith('Player says:'))]
# (~df1['current turn'].str.startswith('MATT:'))
# (~df1['current turn'].str.startswith('DM Responds:'))
# (~df1['current turn'].str.startswith('MATT Responds:')) &
```

```python
df2
```

```python
df3 = df2[(~df2['current turn'].str.startswith('AUDIENCE:'))]
```

```python
df3
```

```python
df4 = df3.drop_duplicates(subset=['context'], inplace=False)
```

```python
df4
```

```python
df5 = df4.drop_duplicates(subset=['current turn'], inplace=False)
```

```python
df5
```

```python
df6 = df5[70:] #Point at which gameplay starts
```

```python
df6
```

```python
df6.to_json('cleaned_data.json', orient='split', compression='infer', index=True)
```
